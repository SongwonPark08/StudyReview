```python
## 문제 
#각 학생의 평균 점수를 계산하여 딕셔너리로 반환하는 함수
def cal_mean(students:list) -> dict:
    mean = 0
    mean_dict = {}
    for i in range(len(students)):
        a = students[i]['subjects'].values()
        a = list(a) 
        b = (a[0] + a[1] + a[2]) / 3
        mean_dict[students[i]["name"]] = round(b, 2)
    return mean_dict
```

```python
cal_mean(students)
```

```python
def cal_subject_average(students:list, subject_name:str.strip()) -> float:
    all = 0 # 모든 과목 합을 받을 변수

    for i in range(len(students)):    
        all += students[i]["subjects"][subject_name]
        
    return float(all / len(students))

```

```python
cal_subject_average(students, "과학")
```
